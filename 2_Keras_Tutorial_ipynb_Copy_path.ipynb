```
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense

import numpy as np

# ==============================
# 1. 產生訓練資料
# ==============================

# 【修改】資料筆數：原本 1000 筆，改為 1500 筆
# 每筆資料包含 10 個隨機產生的浮點數
data = np.random.random((1500, 10))

# 【修改】資料數值範圍：原本乘以 1，改為乘以 2
# 讓每個輸入數值的範圍從 [0, 1) 變成 [0, 2)
data = data * 2

# 【修改】標籤數量：1500 筆，每筆標籤為 0 或 1
labels = np.random.randint(2, size=(1500, 1))

# 印出資料的形狀
print(np.shape(data))

# 印出第一筆資料
print(data[0])

# 印出標籤的形狀
print(np.shape(labels))

# 印出第一筆標籤
print(labels[0])


# ==============================
# 2. 建立神經網路模型
# ==============================

# 建立 Sequential 順序模型
model = Sequential()

# 【修改】隱藏層：32 個神經元，使用 ReLU 激活函數
# input_dim=10 表示每筆輸入資料有 10 個特徵
model.add(Dense(32, activation='relu', input_dim=10))

# 【修改】輸出層：1 個神經元，使用 sigmoid
# 輸出介於 0 到 1，可用於二元分類
model.add(Dense(1, activation='sigmoid'))


# ==============================
# 3. 編譯模型
# ==============================

# 【修改】optimizer：調整模型權重的方法
# 【修改】loss：二元分類使用 binary_crossentropy
# 【修改】metrics：訓練時顯示準確率
model.compile(
    optimizer='rmsprop',
    loss='binary_crossentropy',
    metrics=['accuracy']
)


# ==============================
# 4. 訓練模型
# ==============================

# 【修改】epochs=15：訓練 15 個回合
# 【修改】batch_size=64：每次使用 64 筆資料更新模型
model.fit(
    data,
    labels,
    epochs=15,
    batch_size=64
)


# ==============================
# 5. 評估模型
# ==============================

# 使用訓練資料評估模型的損失與準確率
score = model.evaluate(data, labels)

# 印出模型準確率
print("模型準確率：", score[1])
```

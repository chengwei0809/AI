```
# ==============================
# 1. 匯入所需套件
# ==============================

from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense, Conv2D, MaxPooling2D, Flatten


# ==============================
# 2. 建立 CNN 卷積神經網路模型
# ==============================

# 建立 Sequential 順序模型
model = Sequential()

# 第一層卷積層
# 【已修改】filters：原本 16 改為 32
# 【已修改】kernel_size：原本 5×5 改為 3×3
# input_shape=(28,28,1)：輸入圖片大小為 28×28，1 代表灰階圖片
# activation='relu'：使用 ReLU 激活函數
model.add(Conv2D(
    filters=32,
    kernel_size=(3, 3),
    input_shape=(28, 28, 1),
    activation='relu'
))

# 第一層池化層
# pool_size=(2,2)：將圖片特徵縮小
model.add(MaxPooling2D(pool_size=(2, 2)))


# 第二層卷積層
# 【新增】第二層卷積層，使用 64 個卷積核
# kernel_size=(3,3)：使用 3×3 卷積核
model.add(Conv2D(
    filters=64,
    kernel_size=(3, 3),
    activation='relu'
))

# 第二層池化層
# 將特徵圖進一步縮小
model.add(MaxPooling2D(pool_size=(2, 2)))


# ==============================
# 3. 展平資料
# ==============================

# 將多維特徵圖轉換成一維資料
# 方便傳入後面的全連接層
model.add(Flatten())


# ==============================
# 4. 建立輸出層
# ==============================

# 建立輸出層
# Dense(10)：10 個神經元，分別代表數字 0～9
# softmax：計算每個數字的預測機率
model.add(Dense(10, activation='softmax'))


# ==============================
# 5. 查看模型架構
# ==============================

# 顯示每一層的結構、輸出形狀與參數數量
model.summary()


# ==============================
# 6. 編譯模型
# ==============================

# 【已修改】optimizer：原本 rmsprop 改為 adam
# loss：使用 categorical_crossentropy
# metrics：使用 accuracy 計算準確率
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```
# ==============================
# 1. 載入 MNIST 手寫數字資料集
# ==============================

# 指定使用 Keras 內建的 MNIST 資料集
mnist = keras.datasets.mnist

# 載入資料，分成訓練資料與測試資料
(train_data, train_label), (test_data, test_label) = mnist.load_data()

# ==============================
# 2. 備份原始測試標籤
# ==============================

# 保留尚未轉換的測試標籤
# 後續製作交叉列表（Crosstab）時可以使用
test_label_original = test_label


# ==============================
# 3. 將標籤轉換成 One-hot 編碼
# ==============================

# 【已修改】明確指定 10 個分類（數字 0～9）
# 將訓練標籤轉換成 One-hot 格式
train_label = keras.utils.to_categorical(train_label, num_classes=10)

# 將測試標籤轉換成 One-hot 格式
test_label = keras.utils.to_categorical(test_label, num_classes=10)
```

```
# ==============================
# 4. 訓練 CNN 卷積神經網路模型
# ==============================

# 使用訓練資料開始訓練模型
# train_data：MNIST 手寫數字圖片
# train_label：轉換成 One-hot 格式的正確標籤

# 【已修改】epochs：原本 10 改為 15
# 代表完整訓練資料會被模型學習 15 次

# 【已修改】batch_size：原本 512 改為 256
# 代表每次使用 256 筆資料進行模型訓練

# train_history：儲存每次訓練的歷史紀錄
train_history = model.fit(
    train_data,
    train_label,
    epochs=15,
    batch_size=256
)
```

```
# 使用測試資料評估模型表現
score = model.evaluate(
    test_data,       # 測試圖片
    test_label,      # 測試圖片的 One-hot 標籤
    batch_size=256,  # 每次評估使用 256 筆資料
    verbose=1        # 顯示評估進度
)

# 印出模型的準確率
print("測試資料準確率：", score[1])
```

```
# 使用模型預測測試資料
import pandas as pd
import numpy as np

prediction = model.predict(test_data, batch_size=256)

# 取得每筆資料預測機率最高的類別
prediction_label = np.argmax(prediction, axis=1)

# 顯示原始測試標籤的形狀
print("測試標籤形狀：", test_label_original.shape)

# 建立混淆矩陣
confusion_matrix = pd.crosstab(
    test_label_original,
    prediction_label,
    rownames=['label'],
    colnames=['predict']
)

# 顯示混淆矩陣
print(confusion_matrix)
```

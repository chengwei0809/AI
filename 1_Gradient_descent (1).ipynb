```python
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# 1. 定義目標函數
# ==============================

# 目標函數：y = x²
def func(x):
    return np.square(x)

# 目標函數的一階導數：dy/dx = 2x
def dfunc(x):
    return 2 * x


# ==============================
# 2. 梯度下降法
# ==============================

def GD(x_start, df, epochs, lr):
    """
    梯度下降法：透過反覆更新 x，尋找目標函數的最小值。

    x_start：起始位置
    df：目標函數的一階導數
    epochs：迭代次數
    lr：學習率
    """

    # 建立陣列，記錄每次迭代後的 x 值
    xs = np.zeros(epochs + 1)

    # 設定初始位置
    x = x_start
    xs[0] = x

    # 重複執行梯度下降
    for i in range(epochs):

        # 計算目前位置的梯度
        dx = df(x)

        # 計算更新幅度，負號代表往梯度相反方向移動
        v = -dx * lr

        # 更新 x 的位置
        x += v

        # 記錄更新後的 x 值
        xs[i + 1] = x

    # 回傳所有迭代位置
    return xs


# ==============================
# 3. 設定參數
# ==============================

# 【已修改】起始位置：原本 5，改為 -4
x_start = -4

# 【已修改】迭代次數：原本 15，改為 20
epochs = 20

# 【已修改】學習率：原本 0.3，改為 0.1
lr = 0.1


# ==============================
# 4. 執行梯度下降
# ==============================

# 呼叫 GD 函數，取得每次迭代後的 x 值
x = GD(x_start, dfunc, epochs, lr=lr)

# 印出梯度下降的所有位置
print(x)


# ==============================
# 5. 繪製目標函數與梯度下降路徑
# ==============================

# 【已修改】路徑顏色：原本紅色，改為綠色
color = 'g'

# 【已修改】繪圖範圍：原本 -6 到 6，改為 -5 到 5
# 取樣間距仍為 0.01
t = np.arange(-5.0, 5.0, 0.01)

# 繪製目標函數 y = x²
plt.plot(t, func(t), c='b', label='y=x^2')

# 繪製梯度下降路徑
plt.plot(x, func(x), c=color, label='lr={}'.format(lr))

# 用圓點標示每次迭代的位置
plt.scatter(x, func(x), c=color)

# 顯示圖例
plt.legend()

# 顯示圖表
plt.show()
```
